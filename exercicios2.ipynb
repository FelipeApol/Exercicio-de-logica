### Algoritmo do maior e menor

3. Faça um algoritmo para ler dois números e imprimir o maior, o menor ou então dizer se são iguais.

Algoritmo:

1. Leia o primeiro número e salve na varável nume1
2. Leia o segundo número  e salve na variável num2
3. Descobrir o maior
4. Mostrar o maior
5. Descobrir o menor
6. mostrar o menor
7. Descobrir se são iguais
8. Mostrar se são iguais

```python
num1 = float(input("Digite o primeiro número: "))
num2 = float(input("Digite o segundo número: "))

if num1 > num2:
    print(f"O maior número é {num1}")
    print(f"O menor número é {num2}")
elif num2 < num1:
    print(f"O maior número é {num2}")
    print(f"O menor número é {num1}")
else:
    print("Os números são iguais")
```

# Exercicio 4
Faça um algoritmo para ler um número inteiro e dizer se o número lido é par ou impar.

1. Leia o primeiro número e salve na variável numero.
2. Descobrir se é par ou impar: divindindo por 2 e se tiver sobra ele é impar.
3. Mostre o resultado.


```python
numero = int(input("Digite um número inteiro: "))

if numero % 2 == 0:
    print(f"O número {numero} é par.")
else:
    print(f"O número {numero} é ímpar.")
```

# Exercicio 5

Faça um algoritmo para ler dois números A e B e dizer se A é divisível por B.

1. Leia o primeiro número e salve na varável A
2. Leia o segundo número e salve na variável B
3. mostre se o A é divisívio por B

```python
A = int(input("Digite o primeiro número:"))
B = int(input("Digite o segundo número:"))
if B == 0:
    print("Não é possível dividir por zero.")
else:
    if A % B == 0:
        print(f"{A} é divisível por {B}.")
    else:
        print(f"{A} não é divisível por {B}.")
```

# Exercico 6

Faça um algoritmo para ler dois números e imprimi-los em ordem crescente.
1. Leia o primeiro número e salve na varável n1
2. Leia o segundo número e salve na variável n2
3. descobrir o menor
4. mostrar o maior número
5. mostre o maior em seguida

```python

n1 = float(input("Digite o primeiro número: "))
n2 = float(input("Digite o segundo número: "))

if n1 < n2:
    print(f"{n1}, {n2}")
else:
    print(f"{n2}, {n1}")
```
